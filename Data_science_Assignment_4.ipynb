```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
patient_df = pd.read_csv("/content/Patient_Data.csv")
billing_df = pd.read_csv("/content/Billing_Data.csv")

print("--- Patient Dataset Summary ---")
patient_df.info()
```

```python
# Task 2: Select only the columns relevant for billing
billing_relevant_df = patient_df[
    ["PatientID", "Department", "Doctor", "BillAmount"]
]
```

```python
# Task 3: Drop administrative columns like ['ReceptionistID', 'CheckInTime']
patient_df = patient_df.drop(
    columns=["ReceptionistID", "CheckInTime"], errors="ignore"
)
```

```python
# Task 4: Use groupby to find total bill amount per department
dept_total_bill = patient_df.groupby("Department")["BillAmount"].sum()
print("\n--- Total Bill Amount per Department ---")
print(dept_total_bill)
```

```python
# Task 5: Remove duplicate patient records based on PatientID
patient_df = patient_df.drop_duplicates(subset=["PatientID"])
```

```python
# Task 6: Fill missing BillAmount values with the mean bill amount
mean_bill = patient_df["BillAmount"].mean()
patient_df["BillAmount"] = patient_df["BillAmount"].fillna(mean_bill)
```

```python
# Task 7: Merge the billing dataset with patient dataset on PatientID
merged_df = pd.merge(patient_df, billing_df, on="PatientID", how="left")
```

```python
# Task 8: Concatenate an additional DataFrame that contains new patients for the current week (row-wise)
new_patients = pd.DataFrame(
    [
        {
            "PatientID": 107,
            "Name": "Grace",
            "Department": "Cardiology",
            "Doctor": "Dr. Smith",
            "BillAmount": 5500.0,
            "InsuranceCovered": 2000,
            "FinalAmount": 3500,
        },
        {
            "PatientID": 108,
            "Name": "Hank",
            "Department": "Neurology",
            "Doctor": "Dr. John",
            "BillAmount": 6800.0,
            "InsuranceCovered": 1800,
            "FinalAmount": 5000,
        },
    ]
)

merged_df = pd.concat([merged_df, new_patients], ignore_index=True)
```

```python
# Task 9: Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise)
extra_billing_cols = pd.DataFrame(
    {
        "DiscountApplied": [500] * len(merged_df),
        "PaymentStatus": ["Paid"] * len(merged_df),
    }
)

final_cleaned_df = pd.concat([merged_df, extra_billing_cols], axis=1)

print("\n--- Final Cleaned Dataset ---")
print(final_cleaned_df)
```
